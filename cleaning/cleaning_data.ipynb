```python
df_clean_raw["kecamatan"] = (
    df_clean_raw["alamat"]
    .str.extract(r"Kec\.\s*([^,]+)", expand=False)
    .str.strip()
)

display(
    df_clean_raw[["alamat", "kecamatan"]].head(20)
)
```

```python
df_clean_raw["kelurahan"] = (
    df_clean_raw["alamat"]
    .str.extract(r",\s*([^,]+),\s*Kec\.", expand=False)
    .str.strip()
)

display(
    df_clean_raw[["alamat", "kelurahan"]].head(20)
)
```
